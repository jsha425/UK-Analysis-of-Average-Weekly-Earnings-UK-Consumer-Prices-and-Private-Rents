```python
earnings_clean = earnings.iloc[:,[0,1]].copy()
earnings_clean.columns = ["Date", "Weekly_Earnings"]
earnings_clean.head()
```

```python
earnings_clean["Date"] = pd.to_datetime(
    earnings_clean["Date"],
    errors="coerce"
)
```

```python
earnings_clean = earnings_clean.dropna(subset=["Date"]).copy()
```

```python
earnings_clean["Weekly_Earnings"] = pd.to_numeric(
    earnings_clean["Weekly_Earnings"],
    errors = "coerce"
)
```

```python
earnings_clean = earnings_clean.dropna(
    subset=["Weekly_Earnings"]
).copy()
```

```python
earnings_clean["Year"] = earnings_clean["Date"].dt.year
earnings_clean.head()
```

```python
earnings_clean = earnings_clean[
    (earnings_clean["Year"] >= 2015) &
    (earnings_clean["Year"] <= 2025)
].copy()
```

```python
annual_earnings = (
    earnings_clean
    .groupby("Year")["Weekly_Earnings"]
    .mean()
    .reset_index()
)
annual_earnings
```
