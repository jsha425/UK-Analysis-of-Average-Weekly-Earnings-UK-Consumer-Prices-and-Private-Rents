```python
rent = housing.iloc[:,[0,2,4]].copy()

rent.columns = [
    "Date",
    "Area",
    "Rent_Index"
]
rent.head()
```

```python
rent = rent[rent["Area"] == "United Kingdom"].copy()
```

```python
rent.shape
```

```python
rent["Area"].value_counts()
```

```python
rent["Date"] = pd.to_datetime(
    rent["Date"],
    errors="coerce"
)
rent = rent.dropna(subset=["Date"]).copy()
```

```python
rent["Rent_Index"] = pd.to_numeric(
    rent["Rent_Index"],
    errors = "coerce"
)
rent = rent.dropna(subset = ["Rent_Index"]).copy()
```

```python
rent["Year"] = rent["Date"].dt.year
rent = rent[
    (rent["Year"] >= 2015) &
    (rent["Year"] <= 2025)
].copy()
rent.shape
```

Filtering: to get rid of irrelevant rows and get it to the correct time frame

```python
rent.isna().sum()
```

```python
rent.tail()
```

```python
annual_rent = (
    rent
    .groupby("Year")["Rent_Index"]
    .mean()
    .reset_index()
)
annual_rent
```

```python
rent_2015 = annual_rent.loc[
    annual_rent["Year"] ==2015,
    "Rent_Index"
].iloc[0]
```

```python
annual_rent["Rent_Index_2015"] = (
    annual_rent["Rent_Index"] / rent_2015
) * 100
annual_rent
```

```python
annual_rent["Rent_Change"] = (
    annual_rent["Rent_Index_2015"] - 100
)
```

```python
annual_rent.loc[
    annual_rent["Year"] == 2025
]
```

```python
annual_rent.to_csv(
    "annual_rent_2015_2025.csv",
    index=False
)
```
