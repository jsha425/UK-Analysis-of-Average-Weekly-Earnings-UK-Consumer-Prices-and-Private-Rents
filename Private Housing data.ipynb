```python
rent = housing.iloc[:,[0,2,4]].copy()

rent.columns = [
    "Date",
    "Area",
    "Rent_Index"
]
rent.head()
```

```python
rent = rent[rent["Area"] == "United Kingdom"].copy()
```

```python
rent.shape
```

```python
rent["Area"].value_counts()
```

```python
rent["Date"] = pd.to_datetime(
    rent["Date"],
    errors="coerce"
)
rent = rent.dropna(subset=["Date"]).copy()
```

```python
rent["Rent_Index"] = pd.to_numeric(
    rent["Rent_Index"],
    errors = "coerce"
)
rent = rent.dropna(subset = ["Rent_Index"]).copy()
```

```python
rent["Year"] = rent["Date"].dt.year
rent = rent[
    (rent["Year"] >= 2015) &
    (rent["Year"] <= 2025)
].copy()
rent.shape
```

```python
rent.isna().sum()
```

```python
rent.tail()
```

```python
annual_rent = (
    rent
    .groupby("Year")["Rent_Index"]
    .mean()
    .reset_index()
)
annual_rent
```

```python
rent_2015 = annual_rent.loc[
    annual_rent["Year"] ==2015,
    "Rent_Index"
].iloc[0]
```

```python
annual_rent["Rent_Index_2015"] = (
    annual_rent["Rent_Index"] / rent_2015
) * 100
annual_rent
```

```python
annual_rent["Rent_Change"] = (
    annual_rent["Rent_Index_2015"] - 100
)
```

```python
annual_rent.loc[
    annual_rent["Year"] == 2025
]
```

```python
annual_rent.to_csv(
    "annual_rent_2015_2025.csv",
    index=False
)
```

```python
cpi = pd.read_csv("Cost of Living Analysis.py/UK Cost Of Living Project/Data/Raw/Consumer price inflation tables/clean_cpi.csv")
annual_earnings = pd.read_csv("Cost of Living Analysis.py/UK Cost Of Living Project/Data/Raw/Consumer price inflation tables/weekly_earnings.csv")
```

```python
analysis = cpi.copy()

analysis = analysis.merge(
    annual_rent[[
        "Year",
        "Rent_Index_2015",
        "Rent_Change"
    ]],
    on = "Year",
    how="left"
)

analysis = analysis.merge(
    annual_earnings,
    on="Year",
    how="left"
)
```

```python
print(analysis)
```

```python
plt.figure(figsize=(10,6))

light_blue = "#9ecae1"
mid_blue   = "#4292c6"
dark_navy  = "#084594"

plt.plot(
    analysis["Year"],
    analysis["CPIH_x"],
    marker="o",
    color=light_blue,
    label="CPIH"
)
plt.plot(
    analysis["Year"],
    analysis["Earnings_Index"],
    marker="o",
    color=mid_blue,
    label = "Average Weekly Earnings"
)
plt.plot(
    analysis["Year"],
    analysis["Rent_Index_2015"],
    marker = "o",
    color=dark_navy,
    label = "Private Rent Index"
)

plt.xlabel("Year")
plt.ylabel("Index(2015=100)")
plt.title("UK Price, Earnings and Private Rents, 2015-2025")

plt.legend()
plt.grid(True)
plt.show 
```
